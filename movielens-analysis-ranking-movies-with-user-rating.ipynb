## 🎬 Movie Recommendation Analysis using User Ratings

Understanding user preferences is a key component of building effective recommendation systems. In this project, we analyze the MovieLens dataset to explore how users rate movies and identify the most popular and highly rated films.

The dataset contains user ratings for various movies, along with movie titles and user information. By merging and analyzing these datasets, we gain insights into rating distributions, movie popularity, and overall user behavior.

We perform data aggregation and statistical analysis to compute metrics such as:
- Number of ratings per movie  
- Total rating score  
- Average rating  
- Contribution percentage of each movie  

🎯 **Goal:**  
To identify the most influential and popular movies based on user ratings and uncover patterns that can support recommendation systems.

```python
import pandas as pd
import numpy as np
```

```python
ratings=pd.read_csv('/kaggle/input/datasets/nihalba/u-data/u.data',sep='\t',header=None,names=['user_id','movie_id','rating'],usecols=range(3))
```

```python
ratings
```

```python
movies=pd.read_csv("/kaggle/input/datasets/nihalba/u-item/u.item",encoding="iso-8859-1",sep="|",header=None,names=["movie_id","title"],usecols=range(2))
```

```python
movies
```

```python
user = pd.read_csv('/kaggle/input/datasets/nihalba/u-user/u.user', sep='|', header=None, names=['user_id', 'Age', 'Gender', 'Profession', 'Zipcode'])
```

```python
user
```

```python
rating=pd.merge(movies,ratings)
```

```python
rating
```

```python
rating.head()
```

```python
rating.title.value_counts()
```

```python
mg=rating.groupby('title').agg({'rating':[np.size,np.sum, np.mean]})
```

```python
mg
```

```python
pm=mg.sort_values(('rating','mean'),ascending=False)
```

```python
pm
```

```python
gs=pm['rating']['sum'].sum()
```

```python
pm['percentage']=pm['rating']['sum'].div(gs)*100
```

```python
pm.head()
```

```python
pm['rank']=pm['percentage'].rank(ascending=False)
```

```python
pm.sort_values(by='rank',ascending=True)
```

##  Conclusion

Key takeaways:
